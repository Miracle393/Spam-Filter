```python
import numpy as np
from sklearn.metrics import hamming_loss

y_true = np.array([
    [0, 1],
    [1, 1]
])

y_pred = np.array([
    [1, 1],
    [0, 1]
])

print("Hamming Loss:", hamming_loss(y_true, y_pred))
```

```python
from sklearn.metrics import jaccard_score
```

```python
y_true = np.array([
    [0, 1],
    [1, 1]
])

y_pred = np.array([
    [1, 1],
    [0, 1]
])

print("Jaccard Score:", jaccard_score(
    y_true,
    y_pred,
    average="samples"
))
```
